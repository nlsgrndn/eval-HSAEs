```python
import numpy as np
import pandas as pd
from tqdm import tqdm
import seaborn as sns
import matplotlib.pyplot as plt
import pickle
```

```python
MAIN_METRICS = [
    'ms_score',
    'ms_child_greater_than_parent',
    'avg_intra_cluster_similarity',
    'conditional_activation',
    'fraction_parent_larger_than_child',
    'coverage',
    'conditional_spearman_corr_children_only_max_0',
]
```

```python
import os
from pathlib import Path

# read in all pickle files in directory which are dataframes
# display each dataframe
latex_tables_dir = Path("experiments_results")
experiment_name = "wondact_comparison_nonMC"  # Change this to the desired experiment name_
result_dfs = {}
if latex_tables_dir.exists():
    pickle_files = list(latex_tables_dir.glob("*.pkl"))
    
    for pkl_file in pickle_files:
        if not(experiment_name == pkl_file.stem):
            continue
        print(f"\n{'='*80}")
        print(f"File: {pkl_file.name}")
        print(f"{'='*80}")
        df = pd.read_pickle(pkl_file)
        columns_str = ["config_name"] + MAIN_METRICS
        # only keep columns that contain columns_str
        columns = [col for col in df.columns if any(col_str in col for col_str in columns_str) and (col == "config_name" or "_mean" in col)]
        df = df[columns]
        display(df)
        result_dfs[pkl_file.stem] = df
else:
    print("experiments_results directory not found")
```
