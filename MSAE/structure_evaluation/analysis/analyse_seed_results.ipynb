# Per-Cluster Metric Correlations

Loads the per-cluster metric DataFrames produced by each metric runner and merges them via `load_metrics` (same step `aggregate_single_config_metrics.py` uses internally), but **without aggregation** — one row per cluster, with all metrics as columns. This enables correlating per-cluster `depth_from_root_shortest` with any other per-cluster metric.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from my_config import CONFIGS
from structure_evaluation.clustering_graph_metrics.analysis_intra_hierarchy_metrics_helpers import load_metrics
from path_hub import PathBuilder
# from structure_evaluation.clustering_graph_metrics.compute_hierarchy_graph_acts_metrics import ActsMetricRunner
# from structure_evaluation.clustering_graph_metrics.compute_hierarchy_graph_depth_metrics import DepthMetricRunner
# from structure_evaluation.clustering_graph_metrics.compute_hierarchy_graph_metrics_based_on_feature_hierarchicality_metrics import HierarchicalityMetricRunner
# from structure_evaluation.clustering_graph_metrics.compute_intra_hierarchy_graph_sim_metrics import SimilarityMetricRunner

# CLUSTER_METRICS_CLASSES = [ActsMetricRunner, HierarchicalityMetricRunner, SimilarityMetricRunner, DepthMetricRunner]
```

```python
rows = []
for config_name, config in CONFIGS.items():
    print(f"Config: {config_name}")
    path_builder = PathBuilder(config = config)
    metadata_path = path_builder.get_hierarchical_graph_eval_metadata_file_path()
    metadata_df = pd.read_pickle(metadata_path)
    rows.append(metadata_df)

all_metadata_df = pd.concat(rows, ignore_index=True)
display(all_metadata_df)
```

```python
# export to csv
dino_or_vit = "dino" if "dino" in all_metadata_df["config_name"].iloc[0] else "vit"

all_metadata_df.to_csv(f"all_hierarchical_graph_eval_metadata_{dino_or_vit}.csv", index=False)
```
